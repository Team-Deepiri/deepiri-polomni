```python
try:
    import sympy as sp
    t, tbar = sp.symbols('T Tbar', real=True, positive=True)
    K = -3*sp.log(t + tbar)
    dK_dt = sp.diff(K, t)
    print('dK/dT =', dK_dt)
    assert dK_dt.is_negative
    print('SymPy: Kähler volume term derivative negative — PASS')
except Exception as e:
    print('SymPy skip:', e)
```
