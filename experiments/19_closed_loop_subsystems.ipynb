# 19 — Closed Loop & Path C Subsystems

District simulation → CMB imprint → RBLE scan → feedback, plus inflation D_eff sky patches, Graph-NODE training, PINN Eq.1 loss, GW conductance, and UQE bridge.

```python
from polomni.integration.closed_loop import run_closed_loop
from polomni.integration.workflow import run_lab_workflow
from polomni.core.inflation.sky_patches import deff_fnl_from_simulation
import matplotlib.pyplot as plt

graph, steps = run_closed_loop(steps=3, nside=32, seed=0)
print('axis errors (deg):', [s.axis_error_deg for s in steps])

packets = graph.run_simulation_chain(steps=1, num_choices=4)
inflation = deff_fnl_from_simulation(graph, packets, nside=16)
fnl = inflation['fnl_proxy']
plt.figure(figsize=(8, 3))
plt.plot(fnl[:200])
plt.title('Local f_NL proxy from D_eff sky patches')
plt.ylabel('f_NL proxy')
plt.tight_layout()
plt.show()

wf = run_lab_workflow(target_nside=32, closed_loop_steps=2, use_real_data_pipeline=False)
print('subsystems:', list(wf.subsystems.keys()))
```
